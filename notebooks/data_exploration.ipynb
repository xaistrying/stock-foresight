```python
from vnstock.ui import Reference

ref = Reference()

# source options:
# - 'vci' (richer data, less stable on Colab)
# - 'kbs' (stable default)
df = ref.equity.list_by_exchange(source='vci')

exchanges = ['HSX', 'HNX', 'UPCOM']

type_counts = (
    df[df['exchange'].isin(exchanges)]
    .groupby(['exchange', 'type'])
    .size()
    .unstack(fill_value=0)
)

print(type_counts)
```

```python
from vnstock.ui import Market

mkt = Market()

df = mkt.equity("VIB").ohlcv(start="2018-01-01", end="2026-07-25", count=1990, source='kbs')
print("Start date:", df['time'].min(), "| sessions:", len(df))
```

```python
from vnstock.ui import Reference

ref = Reference()

df = ref.equity.list_by_exchange(source='vci')

hose_stocks = df[(df['exchange'] == 'HSX') & (df['type'] == 'STOCK')]

print("Total number of stocks on HSX/HOSE:", len(hose_stocks))
print(hose_stocks['symbol'].to_list()[:5])
```
